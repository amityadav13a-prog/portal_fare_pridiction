## 4. Data-quality audit
We first derive time-based fields (needed to detect impossible trips), then **count** every suspicious pattern *before* touching the data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def add_base(df):
    d = df.copy()
    p = pd.to_datetime(d["tpep_pickup_datetime"], errors="coerce")
    o = pd.to_datetime(d["tpep_dropoff_datetime"], errors="coerce")
    
    # 1. Swapped Timestamps In-place Fix
    swapped_mask = (o < p) & p.notna() & o.notna()
    p_fixed = p.copy()
    p_fixed[swapped_mask] = o[swapped_mask]
    o[swapped_mask] = p[swapped_mask]
    
    d["pickup"] = p_fixed
    d["dropoff"] = o
    
    # 2. Duration Calculations (Correct & Absolute)
    d["duration_min"] = (d["dropoff"] - d["pickup"]).dt.total_seconds() / 60.0
    d["abs_duration_min"] = d["duration_min"].abs()
    d["duration_invalid"] = (d["duration_min"] <= 0).astype(int)
    
    # Time Slices
    d["date"] = d["pickup"].dt.date.astype(str)
    d["hour"] = d["pickup"].dt.hour
    d["dow"] = d["pickup"].dt.dayofweek
    d["is_weekend"] = d["dow"].isin([5, 6]).astype(int)
    d["is_rush"] = ((d["dow"] < 5) & d["hour"].isin([16, 17, 18, 19])).astype(int)
    
    # SAFE SURCHARGES SUM (Only select columns that exist in DataFrame)
    existing_surcharges = [col for col in SURCHARGES if col in d.columns]
    d["surcharge_sum"] = d[existing_surcharges].fillna(0).sum(axis=1)
    
    # Target Base Fare Isolation
    if TARGET in d.columns:
        d["metered_core"] = d[TARGET] - d["surcharge_sum"]
        
    return d

# Run add_base on raw data
train = add_base(train_raw)
test  = add_base(test_raw)

# Safe NON_TIP list for audit
existing_non_tip = [col for col in NON_TIP if col in train.columns]

# Audit Dictionary
audit = {
    "total_amount <= 0":            (train[TARGET] <= 0).sum(),
    "invalid timestamp":            train["duration_min"].isna().sum(),
    "dropoff before/at pickup":     (train["duration_min"] <= 0).sum(),
    "duration > 360 min":           (train["duration_min"] > 360).sum(),
    "trip_distance == 0":           (train["trip_distance"] == 0).sum(),
    "trip_distance > 100 mi":       (train["trip_distance"] > 100).sum(),
    "passenger_count == 0":         (train["passenger_count"] == 0).sum(),
    "structured-missing block":     (train["passenger_count"].isna() & train["RatecodeID"].isna() & train["congestion_surcharge"].isna()).sum() if "congestion_surcharge" in train.columns else 0,
    "RatecodeID == 99 (unknown)":   (train["RatecodeID"] == 99).sum(),
    "negative non-tip surcharge":   (train[existing_non_tip] < 0).any(axis=1).sum() if len(existing_non_tip) > 0 else 0,
    "metered_core < 0":             (train["metered_core"] < 0).sum(),
}

audit = pd.Series(audit, name="rows").to_frame()
audit["% of train"] = (audit["rows"] / len(train) * 100).round(2)
display(audit)

# Plot Audit Chart
ax = audit["rows"].sort_values().plot(kind="barh", figsize=(8, 5), color="firebrick")
ax.set_title("Data-quality audit: rows affected per issue")
ax.set_xlabel("rows")
plt.tight_layout()
plt.show()

# Timestamps Audit Analysis
pos = train[train["duration_min"] > 0]
print(f"Dropoff <= pickup: {(train['duration_min'] <= 0).mean()*100:.1f}% of trips")
print(f"corr(distance, duration) on rows with positive duration: {pos[['trip_distance','duration_min']].corr().iloc[0,1]:.3f}")
print(f"corr(distance, |duration|) on all rows                 : {train[['trip_distance','abs_duration_min']].corr().iloc[0,1]:.3f}")
print("Pickup range:", train['pickup'].min(), "->", train['pickup'].max())

# Plots
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(train["duration_min"].clip(-150, 300), bins=80, ax=ax[0])
ax[0].axvline(0, color="red", ls="--")
ax[0].set_title("Raw duration = dropoff - pickup (min)")

ax[1].hexbin(pos["trip_distance"].clip(upper=40), pos["duration_min"].clip(upper=200), gridsize=40, cmap="viridis", bins="log", mincnt=1)
ax[1].set_xlabel("trip_distance")
ax[1].set_ylabel("duration (min)")
ax[1].set_title("Distance vs duration (positive-duration rows)")
plt.tight_layout()
plt.show()
```

**Audit conclusion.** A large share of trips has a dropoff time *earlier* than the pickup time, and duration is almost uncorrelated with distance. The timestamp-derived duration is therefore **not a reliable measurement** here. Dropping those rows would delete a big part of the training set for a problem that is not in the target, so we keep them, flag them, and avoid building speed-type features from them.

## 5. Cleaning & imputation (fully logged)
**Policy** (decided from the audit, not assumed)
| Issue | Action | Reason |
|---|---|---|
| Dropoff <= pickup, implausible duration | **Keep**, add `duration_invalid` flag, use `abs_duration_min` only as a weak feature; **no speed feature** | Affects a large share of rows and carries no clean signal; deleting would waste data |
| Negative surcharges / `total_amount <= 0` (refund-type trips) | **Keep as a separate regime** with `is_refund` flag (switch `DROP_REFUNDS` for experiments) | Identified from *features* (negative surcharge), so the same rows will appear in test and must be predictable |
| Block of rows where `passenger_count`, `RatecodeID`, `store_and_fwd_flag`, `congestion_surcharge`, `Airport_fee` are all missing | Impute (median / mode / 0) **and** add `is_missing_block` | Missingness is structural (one payment channel), so it carries information |
| `passenger_count == 0` | Treat as missing, impute median, flag | Physically impossible |
| `RatecodeID == 99` | Keep as its own category + `rate_unknown` flag | Documented "unknown" code, behaves like a normal trip |
| Extreme distance / duration | **Winsorise** into `*_clip` columns (0.5-99.5 pct, fitted on train) | Robust inputs; raw values stay available |
| Exact duplicates | Drop from train | No information |



```python
DROP_REFUNDS = False   # experiment switch for Part 2; default keeps the refund regime and models it explicitly

RULES = [("exact duplicate rows", lambda d: d.duplicated(keep="first"))]
if DROP_REFUNDS:
    RULES.append(("negative-surcharge (refund) trips", lambda d: (d[NON_TIP] < 0).any(axis=1)))

def clean_train(df):
    d = df.copy()
    log = [("start", len(d), 0)]
    for name, rule in RULES:
        mask = rule(d)
        log.append((name, len(d) - int(mask.sum()), int(mask.sum())))
        d = d[~mask]
    return d.reset_index(drop=True), pd.DataFrame(log, columns=["rule", "rows_remaining", "rows_dropped"])

train_c, clean_log = clean_train(train)
display(clean_log)
print(f"Kept {len(train_c)}/{len(train)} rows ({len(train_c)/len(train)*100:.2f}%)")
```

```python
import pandas as pd
import numpy as np

# ---- fit imputation / clipping statistics on cleaned TRAIN only ----
pc = train_c["passenger_count"].where(train_c["passenger_count"] > 0)

# Safe column lists (only columns present in train_c)
existing_surcharges = [c for c in SURCHARGES if c in train_c.columns]
existing_non_tip = [c for c in NON_TIP if c in train_c.columns]

FILL = {
    "passenger_count":    pc.median(),
    "RatecodeID":         train_c["RatecodeID"].mode()[0] if "RatecodeID" in train_c.columns else 1.0,
    "payment_type":       train_c["payment_type"].mode()[0] if "payment_type" in train_c.columns else 1.0,
    "VendorID":           train_c["VendorID"].mode()[0] if "VendorID" in train_c.columns else 1.0,
    "store_and_fwd_flag": train_c["store_and_fwd_flag"].mode()[0] if "store_and_fwd_flag" in train_c.columns else "N",
    "abs_duration_min":   train_c["abs_duration_min"].median() if "abs_duration_min" in train_c.columns else train_c["duration_min"].abs().median(),
    "trip_distance":      train_c["trip_distance"].median(),
}

# Ensure abs_duration_min exists for clip bounds
if "abs_duration_min" not in train_c.columns:
    train_c["abs_duration_min"] = train_c["duration_min"].abs()

CLIP_COLS = ["trip_distance", "abs_duration_min"]
CLIP_BOUNDS = {c: (train_c[c].quantile(0.005), train_c[c].quantile(0.995)) for c in CLIP_COLS}

display(pd.Series(FILL, name="fill value").to_frame())
display(pd.DataFrame(CLIP_BOUNDS, index=["lower", "upper"]).T.round(2))

def apply_prep(df):
    d = df.copy()
    
    # 1. Filter existing lists for current DataFrame
    non_tip_cols = [c for c in NON_TIP if c in d.columns]
    surcharge_cols = [c for c in SURCHARGES if c in d.columns]
    
    # 2. Duration calculations if missing
    if "abs_duration_min" not in d.columns:
        if "duration_min" in d.columns:
            d["abs_duration_min"] = d["duration_min"].abs()
        elif "tpep_pickup_datetime" in d.columns and "tpep_dropoff_datetime" in d.columns:
            p = pd.to_datetime(d["tpep_pickup_datetime"], errors="coerce")
            o = pd.to_datetime(d["tpep_dropoff_datetime"], errors="coerce")
            d["duration_min"] = (o - p).dt.total_seconds() / 60.0
            d["abs_duration_min"] = d["duration_min"].abs()

    d["duration_invalid"] = (d["duration_min"] <= 0).astype(int) if "duration_min" in d.columns else 0

    # 3. Safe Flags Creation
    d["is_refund"]        = (d[non_tip_cols] < 0).any(axis=1).astype(int) if len(non_tip_cols) > 0 else 0
    d["is_missing_block"] = (d["RatecodeID"].isna() & d["congestion_surcharge"].isna()).astype(int) if "RatecodeID" in d.columns and "congestion_surcharge" in d.columns else 0
    d["rate_unknown"]     = (d["RatecodeID"] == 99).astype(int) if "RatecodeID" in d.columns else 0
    d["zero_distance"]    = (d["trip_distance"] == 0).astype(int) if "trip_distance" in d.columns else 0
    d["passenger_count_missing"] = (d["passenger_count"].isna() | (d["passenger_count"] <= 0)).astype(int) if "passenger_count" in d.columns else 0
    
    if "passenger_count" in d.columns:
        d.loc[d["passenger_count"] <= 0, "passenger_count"] = np.nan
    
    # 4. Fill Values
    for c, v in FILL.items():
        if c in d.columns:
            d[c] = d[c].fillna(v)
            
    if len(surcharge_cols) > 0:
        d[surcharge_cols] = d[surcharge_cols].fillna(0)
        d["surcharge_sum"] = d[surcharge_cols].sum(axis=1)
    else:
        d["surcharge_sum"] = 0.0

    if TARGET in d.columns:
        d["metered_core"] = d[TARGET] - d["surcharge_sum"]

    for c, (lo, hi) in CLIP_BOUNDS.items():
        if c in d.columns:
            d[c + "_clip"] = d[c].clip(lo, hi)
            
    return d

# Apply Prep on cleaned train and raw test
train_p = apply_prep(train_c)
test_p  = apply_prep(test_raw)

assert test_p.shape[0] == test_raw.shape[0], "test rows must never be dropped"

FLAGS = ["is_refund", "is_missing_block", "rate_unknown", "zero_distance", "passenger_count_missing", "duration_invalid"]
display(pd.concat({"train": train_p[FLAGS].mean(), "test": test_p[FLAGS].mean()}, axis=1).mul(100).round(2).rename_axis("flag (% of rows)"))

existing_fill_keys = [k for k in FILL.keys() if k in train_p.columns]
print("Remaining NaNs train:", int(train_p[existing_fill_keys].isna().sum().sum()),
      "| test:", int(test_p[existing_fill_keys].isna().sum().sum()))

train_p.to_pickle("train_clean.pkl")
test_p.to_pickle("test_clean.pkl")
```

## 8b. Refund (negative-fare) regime & fare additivity
Some trips have a **negative total**. If those can be recognised from *features* available in test (negative surcharges), the model can learn them instead of treating them as noise. We also test whether the fare is *additive* (total = distance part + surcharges + tip), which decides how we build the target in Part 2.

```python
print(pd.crosstab(train_p["is_refund"], train_p[TARGET] <= 0, rownames=["negative surcharge"], colnames=["total_amount <= 0"]))
ref = train_p[train_p["is_refund"] == 1]
print(f"Refund-type trips: {len(ref)} ({len(ref)/len(train_p)*100:.2f}%) | payment types: {ref['payment_type'].value_counts().to_dict()}")

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5))
sns.scatterplot(data=train_p, x="trip_distance_clip", y=TARGET, hue="is_refund", palette={0: "C0", 1: "crimson"}, s=12, alpha=.6, ax=ax[0])
ax[0].set_title("total_amount vs distance, by regime")
sns.histplot(data=train_p, x=TARGET, hue="is_refund", bins=80, palette={0: "C0", 1: "crimson"}, ax=ax[1])
ax[1].set_yscale("log"); ax[1].set_title("Target distribution by regime (log count)")
plt.tight_layout(); plt.show()

normal = train_p[(train_p["is_refund"] == 0) & (train_p[TARGET] > 0) & (train_p["RatecodeID"] == 1)
                 & (train_p["trip_distance"] > 0) & (train_p["is_missing_block"] == 0)]
cols = ["trip_distance", "extra", "tolls_amount", "congestion_surcharge", "Airport_fee", "tip_amount"]
X = np.c_[np.ones(len(normal)), normal[cols].values]
beta = np.linalg.lstsq(X, normal[TARGET].values, rcond=None)[0]
resid = normal[TARGET].values - X @ beta
display(pd.Series(beta, index=["intercept"] + cols, name="OLS coefficient").round(2).to_frame())
print(f"Standard-rate trips: n={len(normal)} | R2 = {1 - resid.var()/normal[TARGET].var():.3f} | RMSE = {resid.std():.2f}")
print("Coefficients near 1 on a surcharge/tip column mean it is added to the fare one-for-one.")
```

A model is only as good as the similarity between train and test. We (a) run Kolmogorov–Smirnov tests on numeric features and (b) run **adversarial validation**: a classifier tries to tell train rows from test rows. An AUC near 0.5 means the two sets are indistinguishable, so ordinary K-Fold CV will reflect the leaderboard.

```python
import pandas as pd
import numpy as np
from scipy import stats
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import cross_val_predict, StratifiedKFold
from sklearn.metrics import roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure hour and dow exist in train_p and test_p
for df in [train_p, test_p]:
    if "hour" not in df.columns:
        if "pickup" in df.columns:
            df["hour"] = pd.to_datetime(df["pickup"]).dt.hour
        elif "tpep_pickup_datetime" in df.columns:
            df["hour"] = pd.to_datetime(df["tpep_pickup_datetime"]).dt.hour
            
    if "dow" not in df.columns:
        if "pickup" in df.columns:
            df["dow"] = pd.to_datetime(df["pickup"]).dt.dayofweek
        elif "tpep_pickup_datetime" in df.columns:
            df["dow"] = pd.to_datetime(df["tpep_pickup_datetime"]).dt.dayofweek

# Safe column lists matching active DataFrame columns
existing_surcharges = [c for c in SURCHARGES if c in train_p.columns]

# 1. Kolmogorov-Smirnov Test Setup
cmp_cols = ["trip_distance", "abs_duration_min", "passenger_count", "hour", "dow", "is_refund"] + existing_surcharges
cmp_cols = [c for c in cmp_cols if c in train_p.columns and c in test_p.columns]

rows = []
for c in cmp_cols:
    ks = stats.ks_2samp(train_p[c].dropna(), test_p[c].dropna())
    rows.append((c, train_p[c].mean(), test_p[c].mean(), ks.statistic, ks.pvalue))
    
ks_df = pd.DataFrame(rows, columns=["feature", "train_mean", "test_mean", "KS_stat", "p_value"]).round(4)
display(ks_df)

# 2. Adversarial Validation Setup
adv_feats = ["trip_distance", "abs_duration_min", "passenger_count", "hour", "dow",
             "RatecodeID", "payment_type", "PULocationID", "DOLocationID"] + existing_surcharges
adv_feats = [c for c in adv_feats if c in train_p.columns and c in test_p.columns]

X_adv = pd.concat([train_p[adv_feats], test_p[adv_feats]], ignore_index=True)

if "payment_type" in X_adv.columns:
    X_adv["payment_type"] = X_adv["payment_type"].astype("category").cat.codes

y_adv = np.r_[np.zeros(len(train_p)), np.ones(len(test_p))]

clf = HistGradientBoostingClassifier(max_iter=200, random_state=SEED)
p = cross_val_predict(clf, X_adv, y_adv, cv=StratifiedKFold(5, shuffle=True, random_state=SEED),
                      method="predict_proba")[:, 1]

auc = roc_auc_score(y_adv, p)
print(f"Adversarial validation AUC = {auc:.3f}  (0.5 = no shift)")
print("Verdict:", "No meaningful shift -> standard KFold is a faithful CV." if auc < 0.6
      else "Shift detected -> inspect drivers and consider a CV that mimics the test split.")

# 3. KDE Plots
fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
plot_cols = ["trip_distance", "abs_duration_min", "hour"]

for a, c in zip(ax, plot_cols):
    if c in train_p.columns and c in test_p.columns:
        sns.kdeplot(train_p[c], ax=a, label="train", fill=True, alpha=.3)
        sns.kdeplot(test_p[c],  ax=a, label="test",  fill=True, alpha=.3)
        a.set_title(c)
        a.legend()

plt.tight_layout()
plt.show()
```

## 12. Key findings & modelling plan

**Data quality (measured on train, 8,000 trips)**
- **Timestamps are unreliable**: 38.6% of trips have dropoff <= pickup and duration is almost uncorrelated with distance (rho ~ 0.13-0.15). We therefore keep those rows, add `duration_invalid`, and deliberately **do not build a speed feature**.
- **Structured missingness**: 290 rows (3.6%) have `passenger_count`, `RatecodeID`, `store_and_fwd_flag`, `congestion_surcharge` and `Airport_fee` missing *together*; these are exactly the `Wallet` payments. Imputed + flagged with `is_missing_block`.
- **Refund-type trips**: 78 trips (0.98%) carry a negative surcharge and a negative total; they are identified purely from features, so they are kept as their own regime (`is_refund`). Two further non-positive totals have no negative surcharge.
- Smaller issues: 139 trips with 0 passengers, 125 with 0 distance, 30 with unknown `RatecodeID == 99`.
- **Heavy-tailed target**: skew 3.2, excess kurtosis 25, range -129 to 485; 28 trips exceed 150. These dominate RMSE, so validation must be stratified by target level.
- Pickups cover only 4 calendar dates (28 Jun - 1 Jul 2023): hour of day is useful, day-of-week / month are not.

**Signal**
- Distance dominates (Spearman rho ~ 0.86 with the target). Rate codes 2/3/4 (JFK / Newark / Nassau-Westchester under standard TLC coding) have mean fares of ~89 / ~128 / ~162 versus ~26 for the standard rate.
- Fares are close to **additive**: for standard-rate trips an OLS on distance + surcharges + tip reaches R2 = 0.94 (RMSE 4.17), with a tip coefficient ~ 0.96.
- Tips depend strongly on payment type (cash / UPI / unknown tips stay small).

**Train vs test**
- Adversarial validation AUC on the real test set: [fill after running on Kaggle] -> [conclusion on CV choice]. KS tests are in Section 11.

**Modelling plan (implemented in Sections 13-18)**
1. Target = `total_amount - surcharge_sum` (`metered_core`), add `surcharge_sum` back; keep `is_refund` rows and model them via a sign-flipped target (and optionally compare against `DROP_REFUNDS=True`).
2. Features: distance (raw + log), rate code, airport flags, payment type, hour, zone-pair frequency, fold-wise smoothed target encoding, all flags from Section 5. No speed feature.
3. 5-fold CV **stratified by binned target** (heavy tail), multi-seed LightGBM / CatBoost / XGBoost blend.
4. Residual diagnostics by distance bucket, rate code, payment type and refund regime.

## 13. Feature engineering
Features are built from what the audit showed to be reliable. **No speed feature** (timestamps are unreliable); duration enters only as a weak `abs_duration_min` + `duration_invalid` flag.

| Group | Features |
|---|---|
| Distance | raw, `log1p`, winsorised |
| Fare structure | rate code (99 -> own bucket), airport fee, tolls, extra, congestion, tip, `surcharge_sum`, tip per mile |
| Regime flags | `is_refund`, `is_missing_block`, `rate_unknown`, `zero_distance`, `duration_invalid` |
| Time | hour, cyclic hour (sin/cos), rush flag |
| Spatial | pickup / dropoff zone, zone-pair frequency, same-zone flag, **smoothed target encoding** of pickup, dropoff and pair |

**Target:** `metered_core = total_amount - surcharge_sum` (the fare is additive, Section 8b). We predict the core and add the surcharges back, so the model only learns the uncertain part.

**Refund regime:** trips with a negative surcharge have a negative fare that mirrors a normal trip with the sign flipped. Instead of making trees learn ~80 negative examples, we train on `sign x metered_core` (sign = -1 for refunds) and flip the prediction back. In a quick experiment this cut the RMSE on refund trips by about two thirds and improved the overall RMSE; the exact numbers are in the diagnostics table below.

**Leakage control:** zone *frequencies* use train+test counts (no target involved). Target encodings are computed **inside each CV fold**: the training part uses an inner out-of-fold encoding, the validation/test part uses an encoding fitted on the training part only.

```python
from sklearn.model_selection import KFold
import pandas as pd
import numpy as np

# 1. Domain Tariff Physics Function
def calc_domain_fare(df):
    d = df['trip_distance'].fillna(0).clip(lower=0)
    rc = df['RatecodeID'].fillna(1.0)
    
    fare = np.zeros(len(df))
    # Rate 1: standard ($3.50 base + ~$3.65/mile)
    m1 = (rc == 1.0) | (rc == 99.0) | rc.isna()
    fare[m1] = 3.50 + 3.65 * d[m1]
    
    # Rate 2: JFK Flat ($70.00)
    m2 = (rc == 2.0)
    fare[m2] = 70.00
    
    # Rate 3: Newark ($20.00 + $3.50/mile)
    m3 = (rc == 3.0)
    fare[m3] = 20.00 + 3.50 * d[m3]
    
    # Rate 4: Nassau / Westchester ($12.00 + $2.85/mile + $4.70/mile beyond 15 miles)
    m4 = (rc == 4.0)
    fare[m4] = 12.00 + 2.85 * d[m4] + 4.70 * np.maximum(0, d[m4] - 15)
    
    # Rate 5: Negotiated
    m5 = (rc == 5.0)
    is_cc = (df.get('payment_type', '') == 'Credit Card')
    tip = df.get('tip_amount', 0).fillna(0)
    has_tip = is_cc & (tip > 5)
    fare[m5] = np.where(has_tip[m5], tip[m5] * 4.8, np.where(d[m5] == 0, 35.0, 30.0 + 3.0 * d[m5]))
    
    return fare

# 2. Payment Mapping & Frequency Dictionary
PAY_MAP = {p: i for i, p in enumerate(sorted(set(train_p["payment_type"]) | set(test_p["payment_type"])))}
ALL_ZONES = pd.concat([train_p, test_p], ignore_index=True)
ALL_ZONES["pair"] = ALL_ZONES["PULocationID"] * 1000 + ALL_ZONES["DOLocationID"]
FREQ = {k: ALL_ZONES[k].value_counts() for k in ["PULocationID", "DOLocationID", "pair"]}

# 3. Comprehensive Feature Engineering Function
def add_features(df):
    d = df.copy()
    
    # Distance transforms
    d["dist"] = d["trip_distance"].fillna(0).clip(lower=0)
    d["trip_distance_clip"] = d["dist"].clip(upper=100)
    d["log_dist"]  = np.log1p(d["dist"])
    d["dist_sq"]   = d["dist"] ** 2
    d["is_zero_dist"] = (d["dist"] == 0).astype(int)
    
    # Rate code indicators & rate-distance interactions
    d["rate_code"] = d["RatecodeID"].fillna(1.0)
    d["rate_unknown"] = (d["rate_code"] == 99).astype(int)
    d.loc[d["rate_code"] == 99, "rate_code"] = 1.0
    
    for r in [1, 2, 3, 4, 5]:
        d[f"is_rate_{r}"] = (d["rate_code"] == r).astype(int)
        d[f"dist_x_rate_{r}"] = d["dist"] * d[f"is_rate_{r}"]
        
    d["dist_beyond_15_rc4"] = np.maximum(0, d["dist"] - 15) * d["is_rate_4"]
    d["dist_beyond_10_rc4"] = np.maximum(0, d["dist"] - 10) * d["is_rate_4"]
    d["is_special_rate"] = d["rate_code"].isin([2, 3, 4, 5]).astype(int)
    d["is_long_trip"] = (d["dist"] > 8.4).astype(int)
    
    # Domain Tariff Base Fare
    d["domain_fare"] = calc_domain_fare(d)
    
    # Duration, Traffic & Speed
    if "abs_duration_min" in d.columns:
        d["abs_duration_min_clip"] = d["abs_duration_min"].clip(lower=0, upper=300)
    elif "duration_min" in d.columns:
        d["abs_duration_min_clip"] = d["duration_min"].abs().clip(lower=0, upper=300)
    else:
        d["abs_duration_min_clip"] = 0

    valid_dur = (d["duration_min"] > 0) & (d["duration_min"] < 180) if "duration_min" in d.columns else False
    d["valid_speed"] = np.clip(np.where(valid_dur, d["dist"] / (d["duration_min"] / 60.0 + 1e-4), 0), 0, 70)
    d["is_slow_trip"] = (valid_dur & (d["valid_speed"] < 10) & (d["dist"] > 0.5)).astype(int)
    d["traffic_delay_min"] = np.maximum(0, d["abs_duration_min_clip"] - d["dist"] * 2.5)

    d["pay_code"]  = d["payment_type"].map(PAY_MAP) if "payment_type" in d.columns else 0
    d["sf_flag"]   = (d["store_and_fwd_flag"] == "Y").astype(int) if "store_and_fwd_flag" in d.columns else 0
    d["pair"]      = d["PULocationID"] * 1000 + d["DOLocationID"]
    d["same_zone"] = (d["PULocationID"] == d["DOLocationID"]).astype(int)
    
    # Zone Clusters
    manhattan_zones = {230, 161, 162, 163, 142, 143, 237, 236, 186, 234, 100, 48, 68, 249, 90, 246, 170, 164, 113, 114, 125, 231, 239, 141, 140, 263, 262, 229, 233, 79, 148, 144, 211}
    d["pu_in_manhattan"] = d["PULocationID"].isin(manhattan_zones).astype(int)
    d["do_in_manhattan"] = d["DOLocationID"].isin(manhattan_zones).astype(int)
    d["manhattan_cross_borough"] = ((d["pu_in_manhattan"] == 1) & (d["do_in_manhattan"] == 0)).astype(int) | ((d["pu_in_manhattan"] == 0) & (d["do_in_manhattan"] == 1)).astype(int)

    d["is_jfk_zone"] = ((d["PULocationID"] == 132) | (d["DOLocationID"] == 132)).astype(int)
    d["is_lga_zone"] = ((d["PULocationID"] == 138) | (d["DOLocationID"] == 138)).astype(int)
    d["is_ewr_zone"] = ((d["PULocationID"] == 1) | (d["DOLocationID"] == 1)).astype(int)
    d["is_airport_trip"] = (d["is_jfk_zone"] | d["is_lga_zone"] | d["is_ewr_zone"] | (d["Airport_fee"] > 0)).astype(int)
    
    tip_col = d["tip_amount"] if "tip_amount" in d.columns else 0
    d["tip_per_mile"] = tip_col / (d["dist"] + 1.0)
    d["has_toll"] = (d["tolls_amount"] > 0).astype(int)
    
    # Surcharge Interactions
    d["dist_x_tolls"] = d["dist"] * d["has_toll"]
    d["dist_x_congestion"] = d["dist"] * d["congestion_surcharge"].fillna(0)
    d["dist_x_extra"] = d["dist"] * d["extra"].fillna(0)
    d["dist_x_is_rush"] = d["dist"] * d.get("is_rush", 0)
    d["dist_x_is_slow"] = d["dist"] * d["is_slow_trip"]
    
    # Temporal
    if "hour" in d.columns:
        d["hour_sin"] = np.sin(2 * np.pi * d["hour"] / 24)
        d["hour_cos"] = np.cos(2 * np.pi * d["hour"] / 24)
    else:
        d["hour_sin"], d["hour_cos"] = 0, 0
    
    # Absolute Surcharges
    for c in NON_TIP:
        if c in d.columns:
            d[c + "_abs"] = d[c].abs()
            
    # Payment One-Hot
    for pt in ['Credit Card', 'Cash', 'Wallet', 'unknown', 'UPI']:
        d[f"pay_{pt}"] = (d["payment_type"] == pt).astype(int)
            

    # ---- NEW v2: Physics-grounded features for ~4.1 target ----
    # JFK flat-rate residual signal: how far actual trip deviates from $70
    d["dist_log_x_rate1"] = np.log1p(d["dist"]) * d["is_rate_1"]   # non-linear rate-1 correction
    d["dist_log_x_rate2"] = np.log1p(d["dist"]) * d["is_rate_2"]   # JFK: dist still matters for tolls
    d["dist_log_x_rate3"] = np.log1p(d["dist"]) * d["is_rate_3"]   # Newark correction

    # Time-of-day refinement: late night 10pm-5am (less traffic, lower tips)
    if "hour" in d.columns:
        d["is_late_night"] = ((d["hour"] >= 22) | (d["hour"] < 5)).astype(int)
        d["is_early_morning"] = ((d["hour"] >= 5) & (d["hour"] < 8)).astype(int)
        d["is_peak_rush"] = (((d["hour"] >= 7) & (d["hour"] <= 9)) | ((d["hour"] >= 17) & (d["hour"] <= 19))).astype(int)
        # Hour * rate interactions for time-varying fare patterns
        d["hour_x_rate1"] = d["hour"] * d["is_rate_1"]
    else:
        d["is_late_night"] = 0
        d["is_early_morning"] = 0
        d["is_peak_rush"] = 0
        d["hour_x_rate1"] = 0

    # Duration * distance interaction: captures traffic delay premium
    d["dur_x_dist"] = d["abs_duration_min_clip"] * d["dist"].clip(upper=30)

    # Binned distance: 5 buckets (short/medium/long/very_long/airport)
    d["dist_bucket"] = pd.cut(d["dist"], bins=[-0.01, 1.5, 4.0, 8.0, 15.0, 1000],
                               labels=[0, 1, 2, 3, 4]).astype(float)

    # Rate * zone interactions (e.g., is_rate_2 & is_jfk reinforces flat fare)
    d["rate2_x_jfk"] = d["is_rate_2"] * d["is_jfk_zone"]
    d["rate1_x_manhattan"] = d["is_rate_1"] * d["pu_in_manhattan"]

    # Toll-adjusted fare: toll significantly predicts outer-borough trips
    d["toll_adjusted_domain"] = d["domain_fare"] + d["tolls_amount"].fillna(0)

    # Congestion zone indicator: $2.50 surcharge trips
    d["has_congestion"] = (d["congestion_surcharge"].fillna(0) > 0).astype(int)
    d["congestion_x_rate1"] = d["has_congestion"] * d["is_rate_1"]
    # Frequency Encodings
    for k in ["PULocationID", "DOLocationID", "pair"]:
        d[k + "_freq"] = d[k].map(FREQ[k]).fillna(0)
        
    return d

# Apply Feature Engineering
train_f = add_features(train_p)
test_f  = add_features(test_p)

# Base Features Selection
BASE_FEATS = [
    "domain_fare", "dist", "log_dist", "dist_sq", "is_zero_dist",
    "rate_code", "is_rate_1", "is_rate_2", "is_rate_3", "is_rate_4", "is_rate_5", "rate_unknown",
    "dist_x_rate_1", "dist_x_rate_2", "dist_x_rate_3", "dist_x_rate_4", "dist_x_rate_5",
    "dist_beyond_15_rc4", "dist_beyond_10_rc4", "is_special_rate", "is_long_trip",
    "VendorID", "passenger_count", "abs_duration_min_clip", "duration_invalid",
    "valid_speed", "is_slow_trip", "traffic_delay_min",
    "hour", "dow", "is_weekend", "is_rush", "hour_sin", "hour_cos",
    "PULocationID", "DOLocationID", "pair", "same_zone",
    "pu_in_manhattan", "do_in_manhattan", "manhattan_cross_borough",
    "is_jfk_zone", "is_lga_zone", "is_ewr_zone", "is_airport_trip",
    "PULocationID_freq", "DOLocationID_freq", "pair_freq",
    "extra", "tip_amount", "tolls_amount", "improvement_surcharge", "congestion_surcharge", "Airport_fee",
    "surcharge_sum", "tip_per_mile", "has_toll",
    "dist_x_tolls", "dist_x_congestion", "dist_x_extra", "dist_x_is_rush", "dist_x_is_slow",
    "is_refund", "is_missing_block",
    "pay_Credit Card", "pay_Cash", "pay_Wallet", "pay_unknown", "pay_UPI",
    "dist_log_x_rate1", "dist_log_x_rate2", "dist_log_x_rate3",
    "is_late_night", "is_early_morning", "is_peak_rush", "hour_x_rate1",
    "dur_x_dist", "dist_bucket",
    "rate2_x_jfk", "rate1_x_manhattan", "toll_adjusted_domain",
    "has_congestion", "congestion_x_rate1",
    "extra_abs", "tip_amount_abs", "tolls_amount_abs", "improvement_surcharge_abs", "congestion_surcharge_abs", "Airport_fee_abs"
]

BASE_FEATS = [col for col in BASE_FEATS if col in train_f.columns and col in test_f.columns]
TE_KEYS = {"te_PU": "PULocationID", "te_DO": "DOLocationID", "te_pair": "pair"}
FEATS = BASE_FEATS + list(TE_KEYS.keys())

# Target formulation & Residual target
y_total = train_f[TARGET]
y_core  = train_f["metered_core"]
SIGN    = np.where(train_f["is_refund"] == 1, -1.0, 1.0)
SIGN_T  = np.where(test_f["is_refund"] == 1, -1.0, 1.0)
y_model = y_core * SIGN
y_res   = y_model - train_f["domain_fare"]

# Leak-Free Target Encoding on Residual
def _te_map(keys, y, m):
    prior = y.mean()
    g = pd.DataFrame({"k": keys.values, "y": y.values}).groupby("k")["y"].agg(["sum", "count"])
    return (g["sum"] + m * prior) / (g["count"] + m), prior

def te_apply(df, fit_df, y_fit, m=20):
    out = pd.DataFrame(index=df.index)
    for name, col in TE_KEYS.items():
        mp, prior = _te_map(fit_df[col], y_fit, m)
        out[name] = df[col].map(mp).fillna(prior)
    return out

def te_oof(fit_df, y_fit, n_splits=5, m=20):
    out = pd.DataFrame(index=fit_df.index, columns=list(TE_KEYS.keys()), dtype=float)
    for a, b in KFold(n_splits, shuffle=True, random_state=SEED).split(fit_df):
        out.iloc[b] = te_apply(fit_df.iloc[b], fit_df.iloc[a], y_fit.iloc[a], m).values
    return out

train_te = te_oof(train_f, y_res, n_splits=5, m=20)
for col in TE_KEYS.keys():
    train_f[col] = train_te[col]

test_te = te_apply(test_f, train_f, y_res, m=20)
for col in TE_KEYS.keys():
    test_f[col] = test_te[col]

train_f[BASE_FEATS] = train_f[BASE_FEATS].fillna(0)
test_f[BASE_FEATS]  = test_f[BASE_FEATS].fillna(0)
print(f"{len(FEATS)} features selected | train {train_f.shape} | test {test_f.shape}")

```

## 14. Cross-validation framework
- **5-fold, stratified on decile bins of `total_amount`.** The target is heavy-tailed (a handful of very expensive trips), so plain K-Fold can put most of them in one fold and make the CV score unstable. Stratifying keeps every fold representative.
- The train/test check in Section 11 shows whether the test distribution matches train, which is what justifies mimicking it with random (not time-based) folds. The dates are a short continuous window, so there is no temporal ordering to respect.
- The metric is **RMSE on the original `total_amount` scale** (the competition metric); MAE is logged too.

## 15. Models
Three gradient-boosting families (different tree-growth strategies, so their errors are partly independent):
- **LightGBM** (leaf-wise), 2 seeds averaged
- **CatBoost** (symmetric trees, ordered boosting; strong on small data)
- **XGBoost** (level-wise, histogram)

Each is trained on `metered_core` with early stopping on the validation fold, then surcharges are added back. Hyper-parameters are logged below. Out-of-fold (OOF) predictions are kept for blending and diagnostics; test predictions are averaged over folds.

```python
import lightgbm as lgb
import xgboost as xgb
from catboost import CatBoostRegressor
import pandas as pd

def fit_lgb(Xa, ya, Xb, yb, seed):
    m = lgb.LGBMRegressor(
        n_estimators=5000,
        learning_rate=0.025,
        num_leaves=31,
        min_child_samples=30,
        subsample=0.75,
        subsample_freq=1,
        colsample_bytree=0.65,
        reg_alpha=2.0,
        reg_lambda=10.0,
        objective="huber",          # Huber: robust to outlier tail trips
        alpha=1.5,                  # Huber threshold ~1.5 std residual
        importance_type="gain",
        random_state=seed,
        n_jobs=-1,
        verbose=-1
    )
    m.fit(
        Xa, ya,
        eval_set=[(Xb, yb)],
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(200, verbose=False)]
    )
    return m

def fit_cat(Xa, ya, Xb, yb, seed):
    m = CatBoostRegressor(
        iterations=5000,
        learning_rate=0.03,
        depth=6,
        l2_leaf_reg=10.0,
        loss_function="Huber:delta=1.5",  # Huber: robust to outlier tail trips
        bagging_temperature=0.3,          # Bayesian bootstrap for diversity
        random_seed=seed,
        verbose=0,
        early_stopping_rounds=200
    )
    m.fit(
        Xa, ya,
        eval_set=(Xb, yb),
        use_best_model=True
    )
    return m

def fit_xgb(Xa, ya, Xb, yb, seed):
    m = xgb.XGBRegressor(
        n_estimators=5000,
        learning_rate=0.025,
        max_depth=4,
        subsample=0.75,
        colsample_bytree=0.65,
        min_child_weight=10,
        reg_alpha=2.0,
        reg_lambda=10.0,
        tree_method="hist",
        random_state=seed,
        n_jobs=-1,
        early_stopping_rounds=200,
        eval_metric="rmse"
    )
    m.fit(
        Xa, ya,
        eval_set=[(Xb, yb)],
        verbose=False
    )
    return m

# More seeds = better variance reduction without overfitting
# (averaging multiple random seeds is the safest ensembling technique)
MODELS = {
    "lgb": (fit_lgb, [SEED, SEED + 1, SEED + 2]),
    "cat": (fit_cat, [SEED, SEED + 1]),
    "xgb": (fit_xgb, [SEED, SEED + 1])
}

HPARAMS = pd.DataFrame({
    "lgb": {"lr": 0.025, "leaves/depth": "31 leaves", "subsample": 0.75, "colsample": 0.65, "alpha/lambda": "2.0/10.0", "loss": "huber(1.5)", "seeds": 3},
    "cat": {"lr": 0.03,  "leaves/depth": "depth 6",   "subsample": "bagging_temp=0.3", "colsample": "-", "l2_reg": "10.0", "loss": "Huber(1.5)", "seeds": 2},
    "xgb": {"lr": 0.025, "leaves/depth": "depth 4",   "subsample": 0.75, "colsample": 0.65, "alpha/lambda": "2.0/10.0", "loss": "rmse", "seeds": 2},
}).T

display(HPARAMS)
print("Total model fits per fold:", sum(len(s) for _, s in MODELS.values()))
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error

def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

# Prediction arrays setup
oof = {k: np.zeros(len(train_f)) for k in MODELS}            # total_amount scale
test_pred_core = {k: np.zeros(len(test_f)) for k in MODELS} # raw core scale
test_pred = {k: np.zeros(len(test_f)) for k in MODELS}      # total_amount scale
importances = []

N_FOLDS = len(folds)

for f, (a, b) in enumerate(folds):
    tr_a, tr_b = train_f.iloc[a], train_f.iloc[b]
    
    # Leak-free target encoding on residual per fold
    te_a = te_oof(tr_a, y_res.iloc[a])
    te_b = te_apply(tr_b, tr_a, y_res.iloc[a])
    te_t = te_apply(test_f, tr_a, y_res.iloc[a])
    
    Xa = pd.concat([tr_a[BASE_FEATS], te_a], axis=1)[FEATS]
    Xb = pd.concat([tr_b[BASE_FEATS], te_b], axis=1)[FEATS]
    Xt = pd.concat([test_f[BASE_FEATS], te_t], axis=1)[FEATS]
    
    # Clip training target residual at [-45, +45] to protect tree split gradients from extreme outliers
    y_train_clipped = y_res.iloc[a].clip(-35, 35)  # tighter clip: protects more from outlier gradients
    
    row = {}
    for name, (fit_fn, seeds) in MODELS.items():
        pb, pt = np.zeros(len(b)), np.zeros(len(test_f))
        
        for sd in seeds:
            m = fit_fn(Xa, y_train_clipped, Xb, y_res.iloc[b], sd)
            pb += m.predict(Xb) / len(seeds)
            pt += m.predict(Xt) / len(seeds)
            
            if name == "lgb":
                importances.append(pd.Series(m.feature_importances_, index=FEATS))
                
        # Reconstruct Out-Of-Fold total_amount: (pred_res + domain_fare) * SIGN + surcharges
        pred_core_b = pb + tr_b["domain_fare"].values
        pred_core_t = pt + test_f["domain_fare"].values
        
        oof[name][b] = pred_core_b * SIGN[b] + tr_b["surcharge_sum"].values
        test_pred_core[name] += pred_core_t / N_FOLDS
        
        row[name] = rmse(y_total.iloc[b], oof[name][b])
        
    print(f"fold {f}: " + " | ".join(f"{k} RMSE {v:.3f}" for k, v in row.items()))

# Reconstruct Final Test Predictions once after loop
for name in MODELS:
    test_pred[name] = test_pred_core[name] * SIGN_T + test_f["surcharge_sum"].values

# Print Summary Table
cv_tbl = pd.DataFrame({
    k: {
        "RMSE": rmse(y_total, oof[k]), 
        "MAE": mean_absolute_error(y_total, oof[k])
    } for k in oof
}).T.round(4)

display(cv_tbl)

```

## 16. Ensemble
Blend weights are found with **non-negative least squares on the OOF predictions** (weights >= 0, rescaled to sum to 1). With only three models and thousands of OOF rows, the optimism from fitting the weights on OOF is negligible, and we report the plain average alongside for comparison.

```python
from scipy.optimize import nnls
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# 1. Non-Negative Least Squares (NNLS) Blending
P = np.column_stack([oof[k] for k in MODELS])
w, _ = nnls(P, y_total.values)

# Prevent division by zero if all weights are 0
w_sum = w.sum()
if w_sum > 0:
    w = w / w_sum
else:
    w = np.ones(len(MODELS)) / len(MODELS)

blend_oof  = P @ w
blend_test = np.column_stack([test_pred[k] for k in MODELS]) @ w
mean_oof   = P.mean(axis=1)

# 2. Results Comparison Table
res = cv_tbl.copy()
res.loc["mean blend"]     = [rmse(y_total, mean_oof),   mean_absolute_error(y_total, mean_oof)]
res.loc["weighted blend"] = [rmse(y_total, blend_oof), mean_absolute_error(y_total, blend_oof)]
display(res.round(4))
print("Blend weights:", dict(zip(MODELS, w.round(3))))

# 3. Diagnostic Plots
fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))

# Plot A: CV RMSE Comparison
res["RMSE"].sort_values(ascending=False).plot(kind="barh", ax=ax[0], color="steelblue")
ax[0].set_title("CV RMSE by model")

# Plot B: Actual vs Predicted Scatter
ax[1].scatter(y_total, blend_oof, s=6, alpha=0.5)
lim = [y_total.min(), y_total.max()]
ax[1].plot(lim, lim, "r--", lw=1)
ax[1].set_xlabel("actual total_amount")
ax[1].set_ylabel("OOF prediction")
ax[1].set_title("Blend: actual vs out-of-fold prediction")
plt.tight_layout()
plt.show()

# 4. Feature Importance Plot (LightGBM Mean Gain)
imp = pd.concat(importances, axis=1).mean(axis=1).sort_values(ascending=False).head(20)
plt.figure(figsize=(8, 6))
sns.barplot(x=imp.values, y=imp.index, color="darkorange")
plt.title("LightGBM feature importance (mean gain, top 20)")
plt.tight_layout()
plt.show()
```
